## matmul backward test case

```python
import torch
import torch.nn as nn
```

```python

x = torch.tensor([[[ 1.9170,  0.7433, -1.6831, -0.5770],
         [ 0.5002, -0.1631, -0.6377,  0.1790],
         [ 0.8014,  0.9574,  0.5211,  0.6669]],

        [[ 0.0507, -0.7415,  0.7975, -0.1952],
         [-0.0915, -0.2075, -0.6837, -1.4144],
         [-1.0925, -0.9400,  2.0981,  2.5490]]])
weight =  torch.tensor([[ 0.5496, -1.0675, -0.3498, -1.7053, -0.0094],
        [ 0.8118,  0.6951,  1.7514, -0.0827, -0.7742],
        [ 0.6460, -0.5852,  1.0690,  2.4317, -1.1704],
        [ 0.0139, -0.5536, -0.0932,  0.6729, -0.1479]])
bias = torch.tensor([-0.5602, -0.6464, -0.2690,  0.5687, -0.1302])

print(x)
print(weight.transpose(-1,-2))
print(bias)
```

```python
x.requires_grad_(True)
matmul = nn.Linear(4,8)
matmul.weight = nn.Parameter(weight.transpose(-1,-2))
matmul.bias = nn.Parameter(bias)
y = matmul(x)
y.retain_grad()
print(y)
```

```python
loss = y.sum()
```

```python
loss.backward()
print(y.grad)
print(x.grad)
print(matmul.weight.grad.transpose(-1,-2))
print(matmul.bias.grad)
```
